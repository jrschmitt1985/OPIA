### Melhorando o Estilo do Mapa com Mapbox

Para obter um visual de mapa mais detalhado (similar a um "street view" em termos de detalhe de ruas, embora não seja um "street view" interativo 3D), o Plotly usa estilos do Mapbox. O acadêmico pode especificar diferentes `mapbox_style`.

Alguns estilos comuns que não exigem um token de acesso Mapbox são:
*   `"open-street-map"` (padrão)
*   `"carto-positron"` (tema claro, focado em estradas e cidades)
*   `"carto-darkmatter"` (tema escuro)
*   `"stamen-terrain"`
*   `"stamen-toner"`

Para acessar estilos mais avançados e detalhados do Mapbox (como `"mapbox-streets"`, `"mapbox-satellite"`, `"mapbox-satellite-streets"`), você precisará de um [token de acesso Mapbox](https://docs.mapbox.com/help/getting-started/access-tokens/). Uma vez obtido, você pode configurá-lo no Python:

````python
import pandas as pd
import os

# Caminho para o arquivo de dados de produção
producao_path = os.path.join(dados_dir, 'producao_historica_santos.csv')

# Carregar o arquivo CSV em um DataFrame
df_producao = pd.read_csv(producao_path)

print(f"✅ DataFrame df_producao carregado com sucesso!")
display(df_producao.head())
````

````python
# Mapa interativo com Plotly + Mapbox
fig = px.scatter_mapbox(df_pocos_novo,
    lat='Latitude',
    lon='Longitude',
    color='Status_Producao',
    size='Profundidade_Total',
    hover_name='Poco',
    hover_data=['Campo', 'API_Gravity'],
    zoom=8)
# Configuracao do mapa
fig.update_layout(
    mapbox_style="open-street-map", # <<<<<<<<<<<<<<<<<<<<<<<<<<
    title='Campos Petrolíferos - Bacia de Santos',
    height=600)
fig.show()

# --- Gráfico de produção temporal interativo (com df_producao) ---
# Garantir que a coluna 'Data' seja datetime para plots temporais
if 'Data' in df_producao.columns and not pd.api.types.is_datetime64_any_dtype(df_producao['Data']):
    df_producao['Data'] = pd.to_datetime(df_producao['Data'])

fig_producao_temporal = px.line(df_producao,
    x='Data',
    y='Producao_Oleo_bpd',
    color='Poco',
    title='Curvas de Produção de Óleo')
# Adicionando analise de declinio
fig_producao_temporal.add_hline(y=1000, line_dash="dash",
    annotation_text="Limite Econômico")
# Configuracoes avancadas
fig_producao_temporal.update_layout(yaxis_title="Producao (barris/dia)", xaxis_title=
    "Data",
    hovermode='x unified')
fig_producao_temporal.show()

# --- Dashboard multi-painel ---
fig_dashboard = make_subplots(rows=2, cols=2,
    subplot_titles=('Porosidade vs Permeabilidade', 'Produção Temporal', 'Distribuição por Campo', 'Mapa de Localização'),
    specs=[[{'type': 'scatter'}, {'type': 'scatter'}], [{'type': 'bar'}, {'type': 'scattermapbox'}]] )

# 1. Gráfico Porosidade vs Permeabilidade (row=1, col=1)
fig_scatter_poro_perm_temp = px.scatter(df_pocos_novo,
    x='Porosidade_Media',
    y='Permeabilidade_mD',
    color='Campo',
    size='Profundidade_Total',
    hover_name='Poco')
for trace in fig_scatter_poro_perm_temp.data:
    fig_dashboard.add_trace(trace, row=1, col=1)

# 2. Gráfico Produção Temporal (row=1, col=2)
for poco_name in df_producao['Poco'].unique():
    df_poco_subset = df_producao[df_producao['Poco'] == poco_name]
    fig_dashboard.add_trace(go.Scatter(x=df_poco_subset['Data'],
                                       y=df_poco_subset['Producao_Oleo_bpd'],
                                       mode='lines',
                                       name=f'Poco {poco_name}', # Adiciona Poco ao nome para legenda
                                       legendgroup=poco_name,
                                       showlegend=True
                                      ), row=1, col=2)
fig_dashboard.add_hline(y=1000, line_dash="dash", annotation_text="Limite Econômico",
                       row=1, col=2, line_color="gray", annotation_position="bottom right")

# 3. Gráfico Distribuição por Campo (row=2, col=1)
campo_counts = df_pocos_novo['Campo'].value_counts().reset_index(name='Numero_Pocos')
fig_dashboard.add_trace(go.Bar(x=campo_counts['Campo'],
                               y=campo_counts['Numero_Pocos'],
                               name='Poços por Campo',
                               showlegend=False
                              ), row=2, col=1)

# 4. Mapa de Localização (row=2, col=2)
fig_map_temp = px.scatter_mapbox(df_pocos_novo,
    lat='Latitude',
    lon='Longitude',
    color='Status_Producao',
    size='Profundidade_Total',
    hover_name='Poco',
    hover_data=['Campo', 'API_Gravity'],
    zoom=8)
fig_map_temp.update_layout(mapbox_style="open-street-map") #<<<<<<<<<
for trace in fig_map_temp.data:
    fig_dashboard.add_trace(trace, row=2, col=2)

# Atualizar layout do dashboard
fig_dashboard.update_layout(
    height=800,
    title_text='Dashboard Petrofísico Integrado',
    showlegend=True,
    hovermode='closest',
    mapbox=dict(
        style="open-street-map",          #<<<<<<<<<<<<<<<<<<<<<<<<<
        center=dict(lat=df_pocos_novo['Latitude'].mean(), lon=df_pocos_novo['Longitude'].mean()),
        zoom=8
    )
)

# Atualizar títulos dos eixos dos subplots
fig_dashboard.update_xaxes(title_text='Porosidade Média', row=1, col=1)
fig_dashboard.update_yaxes(title_text='Permeabilidade (mD)', row=1, col=1)
fig_dashboard.update_xaxes(title_text='Data', row=1, col=2)
fig_dashboard.update_yaxes(title_text='Produção de Óleo (bpd)', row=1, col=2)
fig_dashboard.update_xaxes(title_text='Campo', row=2, col=1)
fig_dashboard.update_yaxes(title_text='Número de Poços', row=2, col=1)

fig_dashboard.show()

# Tema corporativo personalizado
def apply_petroleum_theme(fig):
    fig.update_layout(template='plotly_white',
        font=dict(family='Arial, sans-serif', size=12),
        colorway=['#1f77b4', '#ff7f0e', '#2ca02c'],
        plot_bgcolor='white',
        paper_bgcolor='white',
        # Titulos e labels
        title_font_size=16,
        title_font_color='#2c3e50',
        # Grades e eixos
        xaxis=dict(showgrid=True, gridcolor='lightgray'),
        yaxis=dict(showgrid=True, gridcolor='lightgray'),
        # Legenda
        legend=dict(bgcolor='rgba(255,255,255,0.8)')
    )
    return fig
````
